Implementation of chain matrix
multiplication using dynamic programming.

```python
import time

def matrix_chain_order(p, n):
    # dp[i][j] stores minimum number of scalar multiplications
    dp = [[0 for _ in range(n)] for _ in range(n)]

    # Length of the matrix chain
    for length in range(2, n):
        for i in range(1, n - length + 1):
            j = i + length - 1
            dp[i][j] = float('inf')

            for k in range(i, j):
                cost = (
                    dp[i][k]
                    + dp[k + 1][j]
                    + p[i - 1] * p[k] * p[j]
                )

                if cost < dp[i][j]:
                    dp[i][j] = cost

    return dp[1][n - 1]


# -------- User Input --------

n = int(input("Enter number of matrices: "))

print("\nEnter dimensions of the matrices:")

dimensions = []

for i in range(n):
    rows = int(input(f"Enter rows of Matrix {i + 1}: "))
    cols = int(input(f"Enter columns of Matrix {i + 1}: "))

    if i > 0 and rows != dimensions[-1]:
        print("Error: Matrix dimensions are not compatible.")
        exit()

    if i == 0:
        dimensions.append(rows)

    dimensions.append(cols)


# -------- Calculate Execution Time --------

start_time = time.perf_counter()

minimum_cost = matrix_chain_order(dimensions, n + 1)

end_time = time.perf_counter()

execution_time = end_time - start_time


# -------- Output --------

print("\nMinimum number of scalar multiplications:", minimum_cost)
print("Execution time: {:.9f} seconds".format(execution_time))

print("\nTime Complexity: O(n^3)")
print("Space Complexity: O(n^2)")
```
