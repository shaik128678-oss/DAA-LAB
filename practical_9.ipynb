implement prim's algorithm

```python
import time

def prim(graph, n):
    selected = [False] * n
    key = [float('inf')] * n
    parent = [-1] * n

    key[0] = 0

    for _ in range(n):
        # Find the vertex with minimum key value
        min_key = float('inf')
        u = -1

        for v in range(n):
            if not selected[v] and key[v] < min_key:
                min_key = key[v]
                u = v

        selected[u] = True

        # Update adjacent vertices
        for v in range(n):
            if graph[u][v] != 0 and not selected[v]:
                if graph[u][v] < key[v]:
                    key[v] = graph[u][v]
                    parent[v] = u

    # Print MST
    total_cost = 0
    print("\nEdges in Minimum Spanning Tree:")

    for i in range(1, n):
        print(f"{parent[i]} -- {i}  Weight = {graph[i][parent[i]]}")
        total_cost += graph[i][parent[i]]

    print("Total MST Cost:", total_cost)


# ---------------- USER INPUT ----------------

n = int(input("Enter number of vertices: "))

graph = []

print("Enter the adjacency matrix:")
print("(Enter 0 if there is no edge)")

for i in range(n):
    row = list(map(int, input(f"Row {i}: ").split()))
    graph.append(row)

# Measure execution time
start_time = time.perf_counter()

prim(graph, n)

end_time = time.perf_counter()

execution_time = end_time - start_time

print(f"\nExecution Time: {execution_time:.8f} seconds")
```

Prim’s Algorithm finds the Minimum Spanning Tree (MST) of a connected weighted graph.
Start from any vertex and repeatedly choose the minimum-weight edge connecting a visited vertex to an unvisited vertex.
Continue until all vertices are included; the selected edges form the MST with minimum total weight.
