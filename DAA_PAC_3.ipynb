```python
# MAX HEAP
# Time Complexity:
#   Insert one element  : O(log n)
#   Delete maximum      : O(log n)
#   Insert n elements   : O(n log n)
#
# Space Complexity:
#   O(n)


import time


class MaxHeap:

    def __init__(self):
        self.heap = []

    # Insert element
    def insert(self, value):
        self.heap.append(value)
        i = len(self.heap) - 1

        while i > 0:
            parent = (i - 1) // 2

            if self.heap[parent] >= self.heap[i]:
                break

            self.heap[parent], self.heap[i] = \
                self.heap[i], self.heap[parent]

            i = parent

    # Delete maximum element
    def delete_max(self):
        if len(self.heap) == 0:
            return None

        maximum = self.heap[0]

        self.heap[0] = self.heap[-1]
        self.heap.pop()

        i = 0

        while True:
            left = 2 * i + 1
            right = 2 * i + 2
            largest = i

            if left < len(self.heap) and \
                    self.heap[left] > self.heap[largest]:
                largest = left

            if right < len(self.heap) and \
                    self.heap[right] > self.heap[largest]:
                largest = right

            if largest == i:
                break

            self.heap[i], self.heap[largest] = \
                self.heap[largest], self.heap[i]

            i = largest

        return maximum


# ---------------- MAIN PROGRAM ----------------

heap = MaxHeap()

n = int(input("Enter number of elements: "))

start_time = time.perf_counter()

for i in range(n):
    value = int(input("Enter element: "))
    heap.insert(value)

end_time = time.perf_counter()

print("\nMax Heap:", heap.heap)

print("Maximum Element:", heap.heap[0])

deleted = heap.delete_max()
print("Deleted Maximum:", deleted)

print("Heap after deletion:", heap.heap)


# ---------------- COMPLEXITY ----------------

execution_time = end_time - start_time

print("\n--- Complexity ---")
print("Time Complexity  : O(n log n)")
print("Space Complexity : O(n)")
print("Execution Time   :", execution_time, "seconds")
```
