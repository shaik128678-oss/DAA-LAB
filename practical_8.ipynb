DFS

```python
import time

# -----------------------------
# Graph implementation
# -----------------------------

class Graph:
    def __init__(self, vertices):
        self.vertices = vertices
        self.graph = {i: [] for i in range(vertices)}

    # Add an undirected edge
    def add_edge(self, u, v):
        self.graph[u].append(v)
        self.graph[v].append(u)

    # DFS using recursion
    def dfs(self, start):
        visited = [False] * self.vertices
        result = []

        def dfs_recursive(vertex):
            visited[vertex] = True
            result.append(vertex)

            for neighbour in self.graph[vertex]:
                if not visited[neighbour]:
                    dfs_recursive(neighbour)

        dfs_recursive(start)
        return result

    # Display graph
    def display(self):
        print("\nGraph:")
        for vertex in self.graph:
            print(vertex, "->", self.graph[vertex])


# -----------------------------
# User Input
# -----------------------------

vertices = int(input("Enter number of vertices: "))
edges = int(input("Enter number of edges: "))

g = Graph(vertices)

print("\nEnter edges (u v):")

for i in range(edges):
    u, v = map(int, input(f"Edge {i + 1}: ").split())
    g.add_edge(u, v)

g.display()

# -----------------------------
# DFS
# -----------------------------

start_vertex = int(input("\nEnter starting vertex for DFS: "))

# Measure execution time
start_time = time.perf_counter()

dfs_result = g.dfs(start_vertex)

end_time = time.perf_counter()

execution_time = end_time - start_time

# -----------------------------
# Output
# -----------------------------

print("\nDFS Traversal:")
print(" -> ".join(map(str, dfs_result)))

print(f"\nExecution time: {execution_time:.9f} seconds")

# -----------------------------
# Complexity
# -----------------------------

print("\nTime Complexity: O(V + E)")
print("Space Complexity: O(V)")
```

```python
import time
from collections import deque


# ---------------------------------
# Graph Implementation
# ---------------------------------

class Graph:
    def __init__(self, vertices):
        self.vertices = vertices
        self.graph = {i: [] for i in range(vertices)}

    # Add an undirected edge
    def add_edge(self, u, v):
        self.graph[u].append(v)
        self.graph[v].append(u)

    # BFS Traversal
    def bfs(self, start):
        visited = [False] * self.vertices
        queue = deque()

        # Start from the given vertex
        visited[start] = True
        queue.append(start)

        result = []

        while queue:
            vertex = queue.popleft()
            result.append(vertex)

            # Visit all adjacent vertices
            for neighbour in self.graph[vertex]:
                if not visited[neighbour]:
                    visited[neighbour] = True
                    queue.append(neighbour)

        return result

    # Display graph
    def display(self):
        print("\nGraph:")
        for vertex in self.graph:
            print(vertex, "->", self.graph[vertex])


# ---------------------------------
# User Input
# ---------------------------------

vertices = int(input("Enter number of vertices: "))
edges = int(input("Enter number of edges: "))

g = Graph(vertices)

print("\nEnter edges (u v):")

for i in range(edges):
    u, v = map(int, input(f"Edge {i + 1}: ").split())
    g.add_edge(u, v)


# Display graph
g.display()


# ---------------------------------
# BFS
# ---------------------------------

start_vertex = int(input("\nEnter starting vertex for BFS: "))

# Start measuring execution time
start_time = time.perf_counter()

bfs_result = g.bfs(start_vertex)

# Stop measuring execution time
end_time = time.perf_counter()

execution_time = end_time - start_time


# ---------------------------------
# Output
# ---------------------------------

print("\nBFS Traversal:")
print(" -> ".join(map(str, bfs_result)))

print(f"\nExecution Time: {execution_time:.9f} seconds")


# ---------------------------------
# Complexity
# ---------------------------------

print("\nTime Complexity: O(V + E)")
print("Space Complexity: O(V)")
```
